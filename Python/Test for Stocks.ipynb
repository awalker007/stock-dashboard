```python
import streamlit as st
import yfinance as yf
import pandas as pd
import datetime

# ----------------------------
# STREAMLIT APP STARTS HERE
# ----------------------------

st.set_page_config(page_title="HAWK", layout="wide")

# Header
st.markdown("<h1 style='text-align: center; color: red;'>HAWK</h1>", unsafe_allow_html=True)

# Sidebar inputs
st.sidebar.header("Stock Input")
ticker = st.sidebar.text_input("Enter Stock Ticker", "AAPL")

# Time range options
time_range = st.sidebar.selectbox(
    "Select Time Range",
    ["1d", "5d", "1mo", "3mo", "6mo", "1y", "5y", "ytd", "max"]
)

# Fetch data
stock = yf.Ticker(ticker)
data = stock.history(period=time_range)

# Display stock name & price
st.subheader(f"{ticker} Stock Data")
st.write(f"Showing data for: **{time_range}**")

# Show chart
st.line_chart(data["Close"])

# Show opening & closing prices if it's a single day
if time_range == "1d" and not data.empty:
    opening_price = data["Open"].iloc[0]
    closing_price = data["Close"].iloc[-1]
    st.metric("Opening Price", f"${opening_price:.2f}")
    st.metric("Closing Price", f"${closing_price:.2f}")

# Show data table
st.write("Raw Data")
st.dataframe(data.tail(10))
```
