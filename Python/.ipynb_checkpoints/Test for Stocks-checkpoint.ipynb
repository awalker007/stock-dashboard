```python
import streamlit as st
import yfinance as yf
import matplotlib.pyplot as plt

# -----------------
# HAWK Dashboard
# -----------------
st.set_page_config(page_title="HAWK", layout="wide")

# Header
st.title("🦅 HAWK")
st.write("Real-time stock monitoring with charts and insights.")

# Sidebar for user inputs
st.sidebar.header("Settings")

# Input ticker symbol
ticker = st.sidebar.text_input("Enter Stock Ticker:", value="AAPL")

# Select time range
time_range = st.sidebar.selectbox(
    "Select Time Range:",
    ("1d", "5d", "1mo", "3mo", "6mo", "1y", "5y", "max"),
    index=2  # default to 1mo
)

# Fetch data
stock = yf.Ticker(ticker)
data = stock.history(period=time_range)

# Check if data is available
if data.empty:
    st.error("No data available for this ticker or time range.")
else:
    # Display stock chart
    st.subheader(f"📊 {ticker} Stock Price - {time_range}")
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(data.index, data["Close"], label="Close Price", linewidth=2)
    ax.set_xlabel("Date")
    ax.set_ylabel("Price (USD)")
    ax.legend()
    ax.grid(True)
    st.pyplot(fig)

    # Show recent data table
    st.subheader("Recent Data")
    st.dataframe(data.tail(10))
```
